**La pregunta:** ¿Qué fracción de la radiación atraviesa una placa de aluminio de 1 a 10 mm, como la de un nanosatélite?

**El verbo:** simular. **Familia de métodos:** simulación de Monte Carlo, comparada con una solución analítica.

El archivo es pequeño: densidad y coeficiente de atenuación a 1 MeV (másico, en cm²/g, y lineal, en cm⁻¹) del aluminio, el plomo y el polietileno, tomados de las tablas del NIST (X-Ray Mass Attenuation Coefficients). La tarea es escribir un Monte Carlo que dispare N fotones de 1 MeV contra una placa de espesor x y cuente cuántos la atraviesan sin interactuar, y compararlo con la solución analítica I/I₀ = e^(−μx).

**Es un modelo de juguete:** solo fotones de una energía, en haz estrecho y sin dispersión. El ambiente real de un nanosatélite es sobre todo de protones y electrones atrapados, que se blindan de otra manera. Aquí aprendemos a verificar una simulación, no a diseñar un blindaje.

*Los parámetros son reales (tablas del NIST); lo que se simula son los fotones. La verdad conocida con que usted audita a la IA es la solución analítica e^(−μx).*
